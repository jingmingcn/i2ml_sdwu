## Solutions

```python
cancer = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    cancer.data, cancer.target, random_state=0)

# Ex2
logreg = LogisticRegression(max_iter=5000).fit(X_train, y_train)
print("logreg  train/test:", round(logreg.score(X_train, y_train), 3),
      round(logreg.score(X_test, y_test), 3))

# Ex3
full = DecisionTreeClassifier(random_state=0).fit(X_train, y_train)
pruned = DecisionTreeClassifier(max_depth=4, random_state=0).fit(X_train, y_train)
print("tree full   train/test:", round(full.score(X_train, y_train), 3),
      round(full.score(X_test, y_test), 3), "<- overfits")
print("tree depth4 train/test:", round(pruned.score(X_train, y_train), 3),
      round(pruned.score(X_test, y_test), 3))

# Ex4
rf = RandomForestClassifier(n_estimators=100, random_state=0).fit(X_train, y_train)
print("random forest test:", round(rf.score(X_test, y_test), 3))
top = np.argsort(rf.feature_importances_)[::-1][:5]
print("top-5 features:", list(cancer.feature_names[top]))

# Ex5
svc_raw = SVC().fit(X_train, y_train)
scaler = StandardScaler().fit(X_train)
svc_scaled = SVC().fit(scaler.transform(X_train), y_train)
print("SVC raw    test:", round(svc_raw.score(X_test, y_test), 3))
print("SVC scaled test:", round(svc_scaled.score(scaler.transform(X_test), y_test), 3))

# Ex6
print("proba[:5]:\n", np.round(rf.predict_proba(X_test[:5]), 3))
```
