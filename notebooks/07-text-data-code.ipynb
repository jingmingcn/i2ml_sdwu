## Solutions

```python
# Ex1
vect = CountVectorizer().fit(texts)
X = vect.transform(texts)
print("Ex1 shape:", X.shape)

# Ex2
clf = LogisticRegression().fit(X, labels)
new = ["a wonderful and enjoyable movie", "a boring and awful waste of time"]
print("Ex2 predictions:", clf.predict(vect.transform(new)))  # -> [1 0]

# Ex3
tfidf = TfidfVectorizer().fit(texts)
X_tfidf = tfidf.transform(texts)
print("Ex3 tfidf shape:", X_tfidf.shape)

# Ex4
names = vect.get_feature_names_out()
coef = clf.coef_[0]
order = np.argsort(coef)
print("Ex4 most negative:", list(names[order[:3]]))
print("Ex4 most positive:", list(names[order[-3:]]))

# Ex5
uni = CountVectorizer().fit(texts)
bi = CountVectorizer(ngram_range=(1, 2)).fit(texts)
print("Ex5 vocab unigram:", len(uni.get_feature_names_out()),
      "| (1,2)-gram:", len(bi.get_feature_names_out()))
```
