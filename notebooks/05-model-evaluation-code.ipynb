## Solutions

```python
# Ex1
iris = load_iris()
scores = cross_val_score(LogisticRegression(max_iter=5000), iris.data, iris.target, cv=5)
print("Ex1 mean CV:", round(scores.mean(), 3))

# Ex2
Xtr, Xte, ytr, yte = train_test_split(iris.data, iris.target, random_state=0)
param_grid = {"C": [0.01, 0.1, 1, 10, 100], "gamma": [0.01, 0.1, 1, 10, 100]}
grid = GridSearchCV(SVC(), param_grid, cv=5).fit(Xtr, ytr)
print("Ex2 best:", grid.best_params_, "CV:", round(grid.best_score_, 3),
      "test:", round(grid.score(Xte, yte), 3))

# Ex3-4
cancer = load_breast_cancer()
Xtr, Xte, ytr, yte = train_test_split(cancer.data, cancer.target, random_state=0)
logreg = LogisticRegression(max_iter=5000).fit(Xtr, ytr)
print("Ex3 confusion matrix:\n", confusion_matrix(yte, logreg.predict(Xte)))
print(classification_report(yte, logreg.predict(Xte),
                            target_names=cancer.target_names))
auc = roc_auc_score(yte, logreg.predict_proba(Xte)[:, 1])
print("Ex4 ROC AUC:", round(auc, 3))
```
