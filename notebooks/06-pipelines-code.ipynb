## Solutions

```python
# Ex1
pipe = Pipeline([("scaler", StandardScaler()), ("svm", SVC())])
pipe.fit(X_train, y_train)
print("Ex1 test:", round(pipe.score(X_test, y_test), 3))

# Ex2
cv_scores = cross_val_score(pipe, X_train, y_train, cv=5)
print("Ex2 mean CV:", round(cv_scores.mean(), 3))

# Ex3
param_grid = {"svm__C": [0.01, 0.1, 1, 10, 100],
              "svm__gamma": [0.01, 0.1, 1, 10, 100]}
grid = GridSearchCV(pipe, param_grid, cv=5).fit(X_train, y_train)
print("Ex3 best:", grid.best_params_, "test:", round(grid.score(X_test, y_test), 3))

# Ex4
mp = make_pipeline(StandardScaler(), SVC()).fit(X_train, y_train)
print("Ex4 steps:", list(mp.named_steps))
print("Ex4 scaler mean_[:3]:", np.round(mp.named_steps["standardscaler"].mean_[:3], 2))
```
